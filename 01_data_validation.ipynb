```python

import pandas as pd
import numpy as np

df = pd.read_csv("project1_algorithmic_pricing_synthetic_dataset.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
# Check missing values in each column
missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)

print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python
print("Total Duplicate Rows:", df.duplicated().sum())
```

```python
# Check column data types
print(df.dtypes)
```

```python
df.info()
```

```python
# Convert date column to datetime format
df["date"] = pd.to_datetime(df["date"], errors="coerce")

# Verify the conversion
print("Date data type:", df["date"].dtype)

print("Invalid dates:", df["date"].isna().sum())

display(df[["date"]].head())
```

```python
# Check numeric columns
numeric_columns = [
    "our_price",
    "competitor_price",
    "units_sold",
    "revenue",
    "marketing_spend"
]

print(df[numeric_columns].dtypes)

print("\nNegative values check:")
print((df[numeric_columns] < 0).sum())
```

```python
# Check region values
print("Regions:")
print(df["region"].value_counts())

# Check treatment values for each region
print("\nTreatment by Region:")
print(df.groupby("region")["treatment"].unique())

# Check period values
print("\nPeriod values:")
print(df["period"].value_counts())

# Check period labels against dates
expected_period = np.where(
    df["date"] >= pd.Timestamp("2025-05-01"),
    "post_treatment",
    "pre_treatment"
)

mismatches = (df["period"] != expected_period).sum()

print("\nPeriod mismatches:", mismatches)
```

```python
# Expected treatment: North after May 1, 2025
expected_treatment = (
    (df["region"] == "North") &
    (df["date"] >= pd.Timestamp("2025-05-01"))
).astype(int)

print(
    "Treatment mismatches:",
    (df["treatment"] != expected_treatment).sum()
)
```

```python
print("===== DATA VALIDATION SUMMARY =====")

print("Total Rows:", len(df))
print("Total Columns:", len(df.columns))
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())
print("Invalid Dates:", df["date"].isna().sum())
print("Treatment Mismatches:", (df["treatment"] != expected_treatment).sum())

print("\nValidation completed.")
```

```python
df.to_csv("project1_algorithmic_pricing_cleaned.csv", index=False)

print("Cleaned CSV saved successfully!")
```
