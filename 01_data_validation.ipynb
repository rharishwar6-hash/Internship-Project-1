```python
import pandas as pd
import numpy as np

# Step 1: Load dataset if df is not already available
df = pd.read_csv("project1_algorithmic_pricing_synthetic_dataset.csv")

# Step 2: Define numeric columns
numeric_cols = [
    "our_price",
    "competitor_price",
    "units_sold",
    "revenue",
    "marketing_spend"
]

# Step 3: Convert date column
df["date"] = pd.to_datetime(df["date"])

# Step 4: Define IQR outlier function
def iqr_outliers(s):
    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    return ((s < q1 - 1.5 * iqr) |
            (s > q3 + 1.5 * iqr)).sum()

# Step 5: Count outliers by region
outliers = df.groupby("region")[numeric_cols].agg(iqr_outliers)

display(outliers)
```

```python

import pandas as pd
import numpy as np

# 1. Define numeric columns
numeric_cols = [
    "our_price",
    "competitor_price",
    "units_sold",
    "revenue",
    "marketing_spend"
]

# 2. Check duplicate date-region-category rows
key_dupes = int(
    df.duplicated(
        subset=["date", "region", "product_category"]
    ).sum()
)

# 3. Check missing days by region
days = df.groupby("region")["date"].agg(["min", "max", "nunique"])
days["expected_days"] = (days["max"] - days["min"]).dt.days + 1
days["missing_days"] = days["expected_days"] - days["nunique"]

# 4. Check revenue against price x units
revenue_mismatch = int(
    ((df["revenue"] - df["our_price"] * df["units_sold"]).abs() > 0.01).sum()
)

# 5. Check treatment-period logic
expected_treatment = (
    (df["region"] == "North") &
    (df["date"] >= pd.Timestamp("2025-05-01"))
).astype(int)

period_mismatch = int(
    (
        (df["treatment"] != expected_treatment) |
        (df["period"] != np.where(
            df["date"] >= pd.Timestamp("2025-05-01"),
            "post_treatment",
            "pre_treatment"
        ))
    ).sum()
)

# 6. Build validation summary
summary = pd.DataFrame({
    "check": [
        "Missing values",
        "Exact duplicate rows",
        "Duplicate date-region-category rows",
        "No missing days",
        "No zero/negative values",
        "Revenue = price x units",
        "Treatment and period logic"
    ],
    "result": [
        df.isna().sum().sum() == 0,
        df.duplicated().sum() == 0,
        key_dupes == 0,
        (days["missing_days"] == 0).all(),
        (df[numeric_cols] > 0).all().all(),
        revenue_mismatch == 0,
        period_mismatch == 0
    ]
})

summary["result"] = summary["result"].map({
    True: "PASS",
    False: "FAIL"
})

display(summary)
```
